```python
# Get all the numbers needed

numbers = [79, 105, 64, 22, 99]
#print(numbers)


#Built in function max:
print(max(numbers))

#Ascending order
Ascending = sorted(numbers)
print(f"Small to big = {Ascending}🔮")

#Descending order
Descending = sorted(numbers, reverse = True)
print(f"Big to small = {Descending}🎐")



#Find the largest number
large_number = numbers[0]
for num in numbers:
    if num > large_number: 
        large_number = num
print(f"The largest number is {large_number}!🌟")

# Find the smallest number
small_number = numbers[0]
for num in numbers:
    if num < small_number:
        small_number = num
print(f"The smallest number is {small_number}!🍬")
```
