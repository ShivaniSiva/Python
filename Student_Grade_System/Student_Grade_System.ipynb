```python
#Write in the file
def write_student():

    num_students = int(input("Enter the number of students to be added"))
    with open ("Student_Grade_System.txt", "w") as file:
        for student in range (num_students):
            name = input("Enter the student's name")
            grade = input("Enter the student's grade")
            result = input("Enter the student's result [pass or fail]")

            file.write(name + ":")
            file.write(grade +":")
            file.write(result)
            file.write("\n")

            print("Students grades has been sucessfully written in the file!✒️")

#Append in the file
def append_student():
    num_students = int(input("Enter the number of students to be added"))
    with open ("Student_Grade_System.txt","a") as file:
        for student in range (num_students):
            name = input("Enter the student's name")
            grade = input("Enter the student's grade")
            result = input("Enter the student's result [pass or fail]")
            
            file.append(name + ":")
            file.append(grade + ":")
            file.append(result)
            file.append("\n")

            print("The students grades has been sucessfully add in the file!💐")

#Read in the file
def read_student():
    with open ("Student_Grade_System.txt", "r") as file:
            
            for line in file:
                clean_text = line.strip()
            
                print(clean_text)
            print("The file has been sucessfully read!📖")

#Starting page
while True:
    
    choice = int(input("""1 - write the file;
                          2 - append the file;
                          3 - read the file;
                          4 - exit the file; Enter a choice"""))

    if choice == 1:
        write_student()

    elif choice == 2:
        append_student()

    elif choice == 3:
        read_student()

    elif choice == 4:
        print("Exiting the program🎐")
        break
```
