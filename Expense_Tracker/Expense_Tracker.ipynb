```python
#Core Data for Expense_Tracker
expense = [
    {
    "title" : "rewe",
    "amount" : 300,
    "category" : "Groceries"
    },
    {
    "title" : "indian store",
    "amount" : 200,
    "category" : "Groceries"
    },
    {
    "title" : "Shivani's education",
    "amount" : 120,
    "category" : "Education"
    },
    {
    "title" : "Kalai's education",
    "amount" : 7500,
    "category" : "Education"
    },
    {
    "title" : "House rent",
    "amount" : 2000,
    "category" : "Rent"
    },
    {
    "title" : "Shiva's tax",
    "amount" : 460,
    "category" : "Taxes"
    },
    {
    "title" : "Kalai's tax",
    "amount" : 370,
    "category" : "Taxes"
    },
    {
    "title" : "Shivani's tax",
    "amount" : 290,
    "category" : "Taxes"
    },
    {
    "title" : "radio tax",
    "amount" : 45,
    "category" : "Taxes"
    },
    {
    "title" : "Eb",
    "amount" : 100,
    "category" : "Monthly bills"
    },
    {
    "title" : "internet",
    "amount" : 40,
    "category" : "Monthly bills"
    }
]

#Main functions

def add_expense():
    num_expenses = int(input("Write how many expenses you want to add"))
    for _ in range (num_expenses):
        title = input("Give the expense a title")
        amount = float(input("Give the amount of the expense which is used"))
        category = input("Give the expense a category")

        expense.append({"title" : title,
                        "amount" : amount,
                        "category" : category})

        print("The expenses have been sucessfully added in the file!")


def view_expense():
    for exp in expense:
        print("Title : ", exp["title"])
        print("Amount : ", exp["amount"])
        print("Category : ", exp["category"])
        print("~~~")


def delete_expense():
    delete = input("Write the title of the expense which needs to be deleted")

    for exp in expense:
        print("Title of the expense:")
        print(exp["title"])
        print("-------------")
        if exp["title"] == delete:
            expense.remove(exp)
            print("The expense is sucessfully deleted!")
            break

def calculate_expense():
    total = 0
    for exp in expense:
        total += exp["amount"]
    print(f"Total expense is ",total)

#Menu
while True:
    
    try:
        choice = int(input("""1 - add expense;
                              2 - view expense;
                              3 - delete expense;
                              4 - show total;
                              5 - exit the programm
                              Enter a choice"""))

    except ValueError:
        print("Please enter a number")
        continue
    
    if choice == 1:
        add_expense()

    elif choice == 2:
        view_expense()

    elif choice == 3:
        delete_expense()

    elif choice == 4:
        calculate_expense()

    elif choice == 5:
        print("Exiting the program")
        break

    else:
        print("Invaild choice")     
```
