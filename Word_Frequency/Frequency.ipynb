```python
#Input
#text = input("Write a text anything")
#special_word = input("Give a word that need to be counted")



#File operation
with open ("Frequency.txt","w") as file:
    file.write("""Cars are fast cars, shiny cars, classic cars, and modern cars that people love to drive.
                Many cars have powerful engines, while other cars are built for comfort, safety, and efficiency,
                  making cars important in daily life. Sports cars, electric cars, family cars, and
                    luxury cars all show how cars continue to evolve and excite car enthusiasts everywhere.""")

# Read the text
with open ("Frequency.txt", "r") as file:
    text = file.read()

# Define dictionary
store = {}

# Count it
for word in text.lower().split():
    #print(word)
    if word in store:
        store[word] += 1
    else:
        store[word] = 1
        
# Answer
for word, count in store.items():
    print(f"{word} : {count}")
```

```python
text = input("Enter a sentence")

#Convert to lower case and split the text into words
words = text.lower().split()

#Add a Dictionary to store the words counts
store = {}

#Condition
for word in words:
#Then here to add up
    if word in store:
        store[word] += 1

#Store for first time here
    else:
        store[word] = 1

#Print the answer
for word, count in store.items():
    print(f"{word} : {count}")
```
