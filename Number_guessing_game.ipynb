```python
# Generate a number between 0 - 100.
import random
secret_number = random.randint(0, 100) 
print(secret_number)


#Get the guessed number and then check the guessed number with the secret number.
while True:
    guessed_number = int(input("Enter any number between 0 and 100 to guess the secret number"))
    if guessed_number > secret_number:
        print("Your guess is high.")

    elif guessed_number < secret_number:
        print("Your guess is low.")

    elif guessed_number == secret_number:
        print("Your guess is right!✌🏻")
        break
    else: 
        print("Give number between 0 - 100")
```
