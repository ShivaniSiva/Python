## **Problem 4: Modulo Understanding**
Compute `18 % 5` and **explain the calculation step-by-step in a comment**.

Your comment should show:
- How many times 5 fits into 18
- What the remainder is and why

**Bonus:** Also compute `10 % 4` and `15 % 5` with brief comments.

````python
##18 % 5
# Step 1 : 18 / 5 = 3 (with leftover)
# Step 2 : 3 * 5 = 15
# Step 3 : 18 - 15 = 3 <- this is modulo

##10 % 4
# Step 1 : 10 / 4 = 2 (with leftover)
# Step 2 : 2 * 4 = 8
# Step 3 : 10 - 8 = 2 <- this is modulo

##15 % 5
# Step 1 : 15 / 5 = 3 (without leftover)
# Step 2 : 3 * 5 = 15
# Step 3 : 15 - 15 = 0 <- this is modulo
````

## **Problem 14: Student Enrollment**
1. Create `age = "19"` (as a string)
2. Create `credits_completed = 120`
3. Convert `age` to an integer.
4. Compute two booleans:
   - `eligible_by_age` = `age >= 18`
   - `eligible_by_credits` = `credits_completed >= 100`
5. Create `eligible_to_enroll` = `eligible_by_age` **and** `eligible_by_credits` *(Note: both conditions must be True)*
6. Print all variables clearly:
```
Age:                  19

```

````python
age = "19"
credits_completed = 120
age = int(age)

eligible_by_age = bool(age >= 18)
eligible_by_credits = bool(credits_completed >= 100)

eligible_to_enroll = bool(eligible_by_age and eligible_by_credits)

print(f"Age:                  {age}")
print(f"Credits completed:    {credits_completed}")
print(f"Eligible by credits:  {eligible_by_age}")
print(f"Eligible to enroll:   {eligible_to_enroll}")
````
